```python
import numpy as np

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense  # each layer is connected to the layers connected to the layer before

x = np.array([
    [2],
    [3],
    [4],
    [5],
    [6],
    [7],
    [8],
    [9]
])

y = np.array([
0,
0,
0,
1,
1,
1,
1,
1
])

model = Sequential()  # layers in sequence

# adding layers:
model.add(Dense(
    8,  # number of hidden layers
    activation='relu',
    input_shape=(1,)  # 1 input
))

model.add(Dense(1, activation='sigmoid'))  # 1 output

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)



model.fit(
    x,
    y,
    epochs=100,  # One epoch = one full pass over all training samples, it passes through all the layers (input, output, hidden)
    verbose=1
)

prediction = model.predict(np.array([[6]]))

print(prediction)
```
