```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

x = [
[1,2],# Coordinates of the first point
[2,1],
[2,2],
[8,8],
[9,8],
[8,9],

]

model = KMeans(n_clusters=2, random_state=42)# 2 clusters
model.fit(x)
print(model.labels_)
plt.scatter(
[x[0] for x in x], # x axis
[x[1] for x in x],# y axis
c=model.labels_
)


plt.show()
```
