```python
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt

players = ["Ahmed", "Mona", "Omar", "Sara", "Hossam", "Nora"]

x = [# speed vs goals
    [90, 3],
    [85, 5],
    [88, 2],
    [92, 7],
    [80, 1],
    [87, 4],
]

model = DBSCAN(
    eps=3,
    min_samples=2
)
labels = model.fit_predict(x)
print(labels)

plt.scatter(
    [point[0] for point in x],
    [point[1] for point in x],
    c=labels,
    cmap="rainbow",# 'rainbow' color pack
    s=100 # dot size
)

for i, name in enumerate(players):
    plt.text(x[i][0] + 0.5, x[i][1] + 0.1, name, fontsize=10, color="black")

plt.xlabel("Speed")
plt.ylabel("Goals")
plt.title("DBSCAN on Player Data")
plt.show()
```
