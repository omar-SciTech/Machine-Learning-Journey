```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

```python


x=[# RAM size vs price
    [1,3000],
    [2,4000],

    [6,15000],
    [8,18000],

    [14,35000],
    [16,40000]

]
```

```python


model = KMeans(n_clusters=3, random_state=42)
model.fit(x)

print(model.labels_)
plt.scatter(
    [x[0] for x in x], 
    [x[1] for x in x],
    c=model.labels_
)

plt.xlabel('RAM size')
plt.ylabel('price')

plt.show()

```
