```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

students = ["Ahmed", "khaled", "Omar", "Sara", "mohamed", "Youssef"]

y = [
    [90, 3],
    [85, 5],
    [88, 2],
    [92, 7],
    [80, 1],
    [95, 4],
]  

pca = PCA(n_components=1)
y_pca = pca.fit_transform(y)
print(y_pca)
print( pca.explained_variance_ratio_)

plt.figure(figsize=(8, 3))
plt.scatter(y_pca[:, 0], [0] * len(y_pca), s=120, color="royalblue")

for i, name in enumerate(students):
    plt.text(y_pca[i, 0] + 0.2, 0.02, name, fontsize=10, color="black")

plt.xlabel("PCA Component 1 (combines degree & study hours)")
plt.show()
```
